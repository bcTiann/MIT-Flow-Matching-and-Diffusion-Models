```python
class MyDate:
    def __init__(self, ):
        self.year = None
        self.month = None
        self.day = None

    def to_string(self, data_str):
        y, m, d = map(int, data_str.split('-')) 
        self.year = y
        self.month = m
        self.day = d


date_str = "2026-10-10"

d1 = MyDate()
d1.to_string(date_str)
print(d1.year)
print(d1.month)
```

```python
class MyDate2:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    @classmethod
    def to_string(cls, data_str):
        y, m, d = map(int, data_str.split('-'))
        return cls(y, m, d)

d2 = MyDate2.to_string("2026-10-10")
d2.year
```
